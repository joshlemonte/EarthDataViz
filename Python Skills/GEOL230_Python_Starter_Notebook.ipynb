# 2. Variables and arithmetic: Python as a geochemistry calculator

A **variable** is a name that points to a value. The `=` sign means *"store the value on the right under the name on the left"* — it is not a math equation.

Good variable names say what the value is **and its units**: `cl_mgL` beats `x` or `data1`. The dataset you will use follows this convention.

*Reading:* Petrelli §2.2 (naming and style) and §2.6 (basic operations); Kong et al. [§2.1 Variables](https://pythonnumericalmethods.studentorg.berkeley.edu/notebooks/chapter02.01-Variables-and-Assignment.html)

````python
# Convert a chloride concentration from mg/L to mmol/L
cl_mgL = 609.95          # chloride in a soil extract (mg/L)
cl_molar_mass = 35.45    # g/mol

cl_mmolL = cl_mgL / cl_molar_mass
print("Cl =", cl_mmolL, "mmol/L")
print(f"Cl = {cl_mmolL:.2f} mmol/L")   # an f-string: the :.2f means '2 decimal places'
````

**Predict →** Seawater has a Na/Cl **molar** ratio of about 0.86. Before running the next cell, decide: will this sample's ratio be above or below seawater? (Na = 2920 mg/L, Cl = 610 mg/L.)

````python
na_mgL = 2920.45
na_molar_mass = 22.99

na_cl_molar = (na_mgL / na_molar_mass) / (cl_mgL / cl_molar_mass)
print(f"Na/Cl molar ratio = {na_cl_molar:.2f}  (seawater is about 0.86)")
````

### Functions
A function is a reusable recipe. It has **inputs** (in the parentheses), does some work, and **returns** an output.
```
def function_name(input1, input2):      <- definition line, ends with a colon
    """What the function does."""         <- docstring: a description for humans
    result = input1 / input2            <- indented body
    return result                       <- the output
```

````python
def mgL_to_mmolL(conc_mgL, molar_mass):
    """Convert a concentration from mg/L to mmol/L."""
    return conc_mgL / molar_mass

# Defining a function does nothing on its own. You have to CALL it:
print(mgL_to_mmolL(609.95, 35.45))

# It also works on a whole column at once:
df["cl_mmolL"] = mgL_to_mmolL(df["cl_mgL"], 35.45)
df[["cl_mgL", "cl_mmolL"]].head(3)
````

### For loops

# 8. Plotting with Matplotlib

There are two styles of Matplotlib code, and AI tools mix them freely:

| "pyplot" style | "object-oriented" style (**use this one**) |
|---|---|
| `plt.scatter(x, y)` | `fig, ax = plt.subplots()` then `ax.scatter(x, y)` |
| `plt.xlabel("...")` | `ax.set_xlabel("...")` |
| `plt.yscale("log")` | `ax.set_yscale("log")` |

Both work. In this course we use the **object-oriented** style: `fig` is the whole figure (the page) and `ax` is one set of axes (the plot area). This style scales up to multi-panel figures and makes AI code easier to read. If an AI gives you `plt.` style code, you can ask it to "rewrite this using fig, ax = plt.subplots()."

*Reading:* Petrelli §3.1 (your first binary diagram) and §4.3 (publication-ready binary diagrams); Kong et al. [§12.1 2D plotting](https://pythonnumericalmethods.studentorg.berkeley.edu/notebooks/chapter12.01-2D-Plotting.html)

````python
fig, ax = plt.subplots(figsize=(6, 4))                    # 1. make an empty figure and axes

ax.scatter(df["dist_coast_km"], df["cl_mgL"],             # 2. x, y data
           s=25, alpha=0.7, edgecolor="k", linewidth=0.3)  #    marker size, transparency, outline

ax.set_yscale("log")                                      # 3. log axis: salts span 3+ orders of magnitude
ax.set_xlabel("Distance to coast (km)")                   # 4. ALWAYS label axes WITH UNITS
ax.set_ylabel("Cl (mg/L)")
ax.set_title("Chloride vs. distance to coast")

fig.tight_layout()                                        # 5. stop labels from being cut off
fig.savefig("cl_vs_distance.png", dpi=300)                # 6. save a high-resolution copy
plt.show()
````

# 10. Reading error messages (the debugging drill)

Errors are normal. Professional programmers see them all day. The skill is reading them.

### How to read a traceback
1. **Start at the bottom line.** It names the error type and gives a short message, e.g. `KeyError: 'Cl_mgL'`.
2. **Then look for the arrow** (`---->`) pointing to the line in *your* cell that failed.
3. Ignore the long middle part (that's inside pandas or Matplotlib) until you need it.

| Error | Usually means |
|---|---|
| `NameError: name 'x' is not defined` | You used a variable (or `pd`, `np`, `plt`) before running the cell that creates it, or you misspelled it |
| `KeyError: 'column'` | That column name doesn't exist — check spelling and capitals against `df.columns` |
| `TypeError: ... 'str' and 'int'` | Mixing text and numbers |
| `SyntaxError` / `IndentationError` | A typo in the code's structure: a missing `)`, `:`, quote, or wrong indent |
| `ParserError`, `HTTPError: 404`, or `FileNotFoundError` | Wrong URL or file path. A GitHub *page* URL loads a web page, not a CSV, and pandas can't parse it. Use the **raw** URL |

### When you ask an AI to fix an error, paste **three** things
1. the code in the cell, 2. the **full** error message, and 3. the output of `df.columns` (or the context cell from Section 9).
